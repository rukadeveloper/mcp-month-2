```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
base_path = r'c:\Users\haruka\Desktop\AI수업\mcp-month-2\data'
```

```python
fish_df = pd.read_csv(os.path.join(base_path, 'fish.csv'))
```

```python
fish_df.Species.unique()
```

```python
fish_df.columns
```

```python
fish_input = fish_df[['Weight', 'Length', 'Diagonal', 'Height', 'Width']].to_numpy()
```

```python
fish_input[:5]
```

```python
fish_target = fish_df['Species'].to_numpy()
```

```python
fish_target
```

```python
from sklearn.model_selection import train_test_split

train_input, test_input, train_target, test_target =\
train_test_split(fish_input, fish_target, random_state=42)
```

```python
print(train_input.shape, test_input.shape)
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()
ss.fit(train_input)
train_scaled = ss.transform(train_input)
test_scaled = ss.transform(test_input)
```

```python
train_input[:5]
```

```python
train_scaled[:5]
```

```python
from sklearn.neighbors import KNeighborsClassifier

kn = KNeighborsClassifier(n_neighbors=3)
kn.fit(train_scaled, train_target);
```

```python
print(kn.score(train_scaled, train_target))
print(kn.score(test_scaled, test_target))
```

```python
kn.predict(test_scaled[:5])
```

```python
proba = kn.predict_proba(test_scaled[:5])

print(kn.classes_)

print(np.round(proba, decimals=4))
```

```python
distances, indexes = kn.kneighbors(test_scaled[3:4])

print(train_target[indexes])
```

```python
bream_smelt_indexes = (train_target == 'Bream') | (train_target == 'Smelt')
```

```python
bream_smelt_indexes
```

```python
np.sum(bream_smelt_indexes)
```

```python
train_bream_smelt = train_scaled[bream_smelt_indexes] # (33, 5)
target_bream_smelt = train_target[bream_smelt_indexes] # (33,)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
lr.fit(train_bream_smelt, target_bream_smelt);
```

```python
print("예측: ", lr.predict(train_bream_smelt[:5]))
```

```python
print("실제: ", target_bream_smelt[:5])
```

```python
print(lr.classes_)
print(lr.predict_proba(train_bream_smelt[:5]))
```

```python
print(lr.coef_, lr.intercept_)
```

```python
decisions = lr.decision_function(train_bream_smelt[:5])
decisions
```

```python
from scipy.special import expit
print(expit(decisions))
```

```python
lr = LogisticRegression(C=20, max_iter=1000) # 규제는 완화, 반복횟수를 늘림
lr.fit(train_scaled, train_target);
```

```python
print(lr.score(train_scaled, train_target))
print(lr.score(test_scaled, test_target))
```
